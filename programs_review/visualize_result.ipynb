# E0：内側ループの開始点の影響の可視化

`programs/analysis_e0.py` の関数を用いて，`outputs/` 以下の結果から図と表を作成する．可視化する実験と手法は下のセルの変数で選択する．

```python
import os, sys
sys.path.insert(0, os.path.abspath('programs'))
import analysis_e0 as a
from IPython.display import Image, Markdown, display

# 可視化する実験（analysis_e0.PROBLEMS のキー）と手法を指定する
EXPERIMENTS = list(a.PROBLEMS)
METHODS = ['SVRG', 'NFG_SVRG', 'ASAI_SVRG']
```

```python
for experiment in EXPERIMENTS:
    runs = a.load_runs(os.path.join(a.OUTPUTS_DIR, experiment))
    if not runs:
        print(f'{experiment}: 結果なし')
        continue
    display(Markdown(f'## {a.PROBLEMS[experiment]["title"]}'))
    display(Markdown(a.summary_table(experiment, runs, METHODS)))
    display(Markdown(a.paired_difference_table(experiment, runs, METHODS)))
    for path in a.plot_problem(experiment, runs, METHODS):
        display(Image(path))
```

```python
# 全ての実験の図と表を outputs/e0_summary.md に保存する
print(a.write_summary())
```
